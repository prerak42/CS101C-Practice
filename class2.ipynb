```python
#function listing all the factors of a given integer in ascending order

def func(val):
  listVal = []
  for i in range(1, val + 1):
    if(val%i==0):
      listVal.append(i)
  return listVal

num = int(input("Enter a number: "))
print(func(num))
```

```python
#function finding the nth number in the Fibonacci Sequence

def Fibonacci(n):

  if n == 0:
    return 0
  if n == 1:
    return 1

  a = 0
  b = 1

  for i in range(2, n + 1):
    a = b
    b = a + b

  return b

val = int(input("Enter n: "))
print(Fibonacci(val))
```
